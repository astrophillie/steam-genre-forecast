## 1. Compare distant-month errors

The table reports MAE for leads 16–27 in average concurrent activity units. Compare models within a category because category scales differ; lower values indicate smaller errors.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
scores = pd.read_csv(ROOT / "reports/statistical_models/scores.csv")
long_lead = scores.loc[scores.window.eq("Leads 16 to 27")]
comparison = long_lead.pivot(index="category", columns="model", values="mae").round(2)
comparison
```

## Interpretation

Last value has the lowest MAE for four categories. Damped seasonal smoothing improves Indie by about 11.3%, while additional components worsen the other category forecasts. The results support keeping the simple baselines in contention. SES gives almost the same errors as last value in four categories.
