## Comparison with the existing forecasts

The search varies tree depth and tree count across four configurations. Each is evaluated over two earlier 27-month windows, using leads 16–27 for selection. The chosen configuration is refitted at the outer cutoff and forecasts the full horizon without future observed inputs. The table shows long-lead mean absolute error, worst-cutoff error and mean signed error.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
comparison = pd.read_csv(ROOT / "reports/tuning/comparison.csv")
selected_view = comparison.loc[comparison.model.isin(["Last value", "Gradient boosting", "Chronologically tuned boosting"])].round(2)
selected_view
```

RPG benefits from tuning: MAE falls to 49,989.67 and its worst-cutoff error is slightly below last value. The almost zero mean signed error reflects balancing positive and negative errors; it does not imply that individual forecasts are close.

Indie and Strategy do not benefit from the search. Their tuned models perform worse than the simpler ML configurations, and both exceed last-value error. These results support keeping the untuned candidates in contention rather than treating tuning as an automatic upgrade.
