## Historical error ranges

An empirical range was tested using only earlier forecast errors whose target months were already observed at each cutoff. At least twelve calibration rows were required. The radius uses the 90th percentile of absolute long-lead errors, with the lower bound clipped at zero.

Only one origin had sufficient earlier evidence. Observed coverage is therefore based on twelve target months per category, and the folds have already informed development.

```python
coverage = pd.read_csv(ROOT / "reports/forecast_uncertainty/coverage.csv")
coverage.round(2)
```

Massively Multiplayer has only 50% coverage in this check, showing that this rule cannot be presented as a reliable 90% prediction interval. High observed coverage for the other categories is also based on too few periods to establish reliability.

## Interpretation of the v1 experiment

The analysis supports reproducible point projections and reveals meaningful model sensitivity. It does not establish dependable interval coverage or a current whole-market forecast. The historical cohort, September 2025 cutoff and unavailable future releases remain material limits.
